# Bilingual Boundary Lab — first look

This notebook presents the canonical Unicode run visualization and deliberately simple word/character n-gram baselines.

```python
from pathlib import Path
import sys

repo_root = Path.cwd().resolve().parent
sys.path.insert(0, str(repo_root / 'src'))

from bbl.normalization import normalized_tokens
from bbl.runs import build_runs
from bbl.visualization import plot_runs
from bbl.ngram import NGramModel
```

```python
corpus_path = (
    repo_root
    / 'data/raw/war_and_peace_ru_fr_wikisource.txt'
)
text = corpus_path.read_text(encoding='utf-8')
words = normalized_tokens(text)
runs = build_runs(text)
len(text), len(words), len(runs)
```

```python
plot_runs(runs, base=10)
```

```python
model = NGramModel[str](n=2).fit(words)
sample = model.generate(
    100,
    seed=words[:2],
    random_seed=1729,
)
print(' '.join(sample))
```
